```python
import os
import json
import spacy
from spacytextblob.spacytextblob import SpacyTextBlob
```

```python
nlp = spacy.load("en_core_web_lg")
```

```python
# Add TextBlob-based sentiment component
nlp.add_pipe("spacytextblob")
```

```python
# Input file path
file_path = "28Jun22.jl"

# Create output file path with 'processed' before extension
base, ext = os.path.splitext(file_path)
output_path = f"{base}_processed{ext}"
```

```python
# Open and load the JSON file
with open(file_path, 'r', encoding='utf-8') as infile, \
     open(output_path, 'w', encoding='utf-8') as outfile:
    for line_number, line in enumerate(infile, start=1):
        try:
            data = json.loads(line)  # Parse JSON object
            data['entities'] = {
                'loc': [],
                'org': [],
                'person': []
            }
            doc = nlp(data['text'])
            for ent in doc.ents:
                entity_type = ent.label_.lower()
                entity_name = ent.text.lower().replace('\n', ' ').strip()
                if entity_type in ['loc', 'org', 'person']:
                    # Check we add this entity in data
                    if entity_name in data['entities'][entity_type]:
                        continue
                    else:
                        data['entities'][entity_type].append(entity_name)
                # Sentiment Analysis
                data['score'] = round(doc._.blob.polarity, 3)
                data['label'] = "positive" if score > 0 else ("negative" if score < 0 else "neutral")
    
            # Save processed data to output JSONL
            outfile.write(json.dumps(data) + '\n')
        except json.JSONDecodeError as e:
            print(f"Error decoding JSON at line {line_number}: {e}")
```
